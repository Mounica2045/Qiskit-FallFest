 Two qubits can be prepared in a state where their measurement outcomes are strongly
correlated. Your task is to modify a two-qubit circuit and investigate whether you can change the relationship
between the two measurement outcomes.

```python
%pip install qiskit qiskit-aer matplotlib
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# 1. Create simulator
simulator = AerSimulator()
SHOTS = 100

# 2. Original Bell-state circuit
original = QuantumCircuit(2, 2)

original.h(0)
original.cx(0, 1)
original.measure([0, 1], [0, 1])

# 3. Modified Bell-state circuit
# Add only ONE gate: X on qubit 1
modified = QuantumCircuit(2, 2)

modified.h(0)
modified.cx(0, 1)
modified.x(1)
modified.measure([0, 1], [0, 1])

# 4. Predictions BEFORE execution
print("PREDICTIONS BEFORE EXECUTION")
print("Original: 00 and 11 should occur about 50% each.")
print("Modified: 01 and 10 should occur about 50% each.")
print("Original outcomes should match.")
print("Modified outcomes should be opposite.")

# 5. Display circuit diagrams
print("\nORIGINAL BELL-STATE CIRCUIT")
print(original.draw(output="text"))

print("\nMODIFIED BELL-STATE CIRCUIT")
print(modified.draw(output="text"))

# 6. Execute both circuits for 100 shots
circuits = [original, modified]
names = ["Original Bell State", "Modified Bell State"]
results = []

print("\nMEASUREMENT RESULTS")

for name, circuit in zip(names, circuits):
    job = simulator.run(circuit, shots=SHOTS)
    counts = job.result().get_counts()
    results.append(counts)

    print(f"\n{name}")
    print("Measurement counts:", counts)
    print("Total shots:", sum(counts.values()))

    for outcome in ["00", "01", "10", "11"]:
        count = counts.get(outcome, 0)
        print(f"{outcome}: {count} shots ({count}%)")

# 7. Plot histograms for comparison
plot_histogram(
    results,
    legend=names,
    title="Bell-State Correlations: 100 Shots Each"
)
plt.tight_layout()
plt.show()

# 8. Compare actual results
print("\nCOMPARISON")

original_counts = results[0]
modified_counts = results[1]

print("Original circuit:")
print("Matching outcomes (00 and 11):",
      original_counts.get("00", 0) +
      original_counts.get("11", 0))

print("Modified circuit:")
print("Opposite outcomes (01 and 10):",
      modified_counts.get("01", 0) +
      modified_counts.get("10", 0))

# 9. Conclusion
print("\nCONCLUSION")
print("The original Bell circuit produces matching outcomes.")
print("The added X gate flips qubit 1.")
print("The modified circuit therefore produces opposite outcomes.")
print("The qubits remain entangled, but their correlation changes.")
```
