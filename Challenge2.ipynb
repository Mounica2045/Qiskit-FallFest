A Hadamard gate creates an equal superposition of |0⟩ and |1⟩, producing a result
similar to a fair quantum coin. But can you control the measurement probabilities? Your task is to modify a
single-qubit circuit so that the measurement outcomes are not equally distributed.

```python

import math
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# 1. Create simulator
simulator = AerSimulator()
SHOTS = 100

# 2. Circuit A: More zeros (RY 60 degrees)
circuit_A = QuantumCircuit(1, 1)
circuit_A.ry(math.pi / 3, 0)
circuit_A.measure(0, 0)

# 3. Circuit B: More ones (RY 120 degrees)
circuit_B = QuantumCircuit(1, 1)
circuit_B.ry(2 * math.pi / 3, 0)
circuit_B.measure(0, 0)

# 4. Circuit C: Equal probabilities (RY 90 degrees)
circuit_C = QuantumCircuit(1, 1)
circuit_C.ry(math.pi / 2, 0)
circuit_C.measure(0, 0)

circuits = [circuit_A, circuit_B, circuit_C]
names = [
    "Circuit A (RY 60 degrees)",
    "Circuit B (RY 120 degrees)",
    "Circuit C (RY 90 degrees)"
]

# 5. Predictions BEFORE execution
predictions = [
    "0: 75%, 1: 25%",
    "0: 25%, 1: 75%",
    "0: 50%, 1: 50%"
]

print("PREDICTIONS BEFORE EXECUTION")
for name, prediction in zip(names, predictions):
    print(name, "->", prediction)

# 6. Display circuit diagrams
print("\nQUANTUM CIRCUITS")
for name, circuit in zip(names, circuits):
    print("\n", name)
    print(circuit.draw(output="text"))

# 7. Execute each circuit for 100 shots
results = []

print("\nACTUAL MEASUREMENT RESULTS")

for name, circuit, prediction in zip(
    names, circuits, predictions
):
    job = simulator.run(circuit, shots=SHOTS)
    counts = job.result().get_counts()
    results.append(counts)

    zeros = counts.get("0", 0)
    ones = counts.get("1", 0)

    print("\n", name)
    print("Prediction:", prediction)
    print("Actual counts:", counts)
    print(f"Number of zeros: {zeros} ({zeros}%)")
    print(f"Number of ones: {ones} ({ones}%)")
    print("Total shots:", sum(counts.values()))

# 8. Display histograms
plot_histogram(
    results,
    legend=names,
    title="Biased Quantum Coin - 100 Shots"
)
plt.tight_layout()
plt.show()

# 9. Compare predictions with actual results
print("\nCOMPARISON")

for name, counts, prediction in zip(
    names, results, predictions
):
    zeros = counts.get("0", 0)
    ones = counts.get("1", 0)

    print(f"\n{name}")
    print("Predicted:", prediction)
    print(f"Observed: {zeros} zeros and {ones} ones")

print("\nCONCLUSION")
print("Circuit A favours outcome 0.")
print("Circuit B favours outcome 1.")
print("Circuit C produces approximately equal probabilities.")
print("Changing the rotation angle changes the measurement probabilities.")

```

```python

print("""
Answer:
Quantum gates change the quantum state and probability amplitudes
of a qubit. When the qubit is measured, the probability of each
outcome is the squared magnitude of its probability amplitude.

For an RY(theta) gate applied to |0>:

P(0) = cos^2(theta/2)
P(1) = sin^2(theta/2)

Changing the rotation angle changes these probabilities.
An angle below 90 degrees favours 0, an angle above 90 degrees
favours 1, and an angle of 90 degrees gives equal probabilities.

Therefore, quantum gates allow us to control the measurement
distribution of a qubit.
""")
```
