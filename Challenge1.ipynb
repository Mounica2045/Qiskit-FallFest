Problem Statement - A quantum circuit can produce different measurement outcomes depending on the
gates applied to a qubit.
Your task is to predict the output of simple quantum circuits before executing them and then compare your
prediction with the actual measurement results.

```python

%pip install qiskit qiskit-aer matplotlib
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# 1. Create the simulator
simulator = AerSimulator()
SHOTS = 100

# 2. Construct three separate circuits.
# Each circuit uses exactly one qubit and one classical bit for measurement.

# Circuit A: one X gate
circuit_A = QuantumCircuit(1, 1)
circuit_A.x(0)
circuit_A.measure(0, 0)

# Circuit B: two X gates
circuit_B = QuantumCircuit(1, 1)
circuit_B.x(0)
circuit_B.x(0)
circuit_B.measure(0, 0)

# Circuit C: one Hadamard gate
circuit_C = QuantumCircuit(1, 1)
circuit_C.h(0)
circuit_C.measure(0, 0)

circuits = [circuit_A, circuit_B, circuit_C]
names = ["Circuit A (X)", "Circuit B (X, X)", "Circuit C (H)"]

# 3. State predictions BEFORE executing any circuits.
predictions = {
    "Circuit A (X)": "100% outcome 1",
    "Circuit B (X, X)": "100% outcome 0",
    "Circuit C (H)": "50% outcome 0 and 50% outcome 1 (probabilistically)",
}

print("PREDICTIONS BEFORE EXECUTION")
for name in names:
    print(f"{name}: {predictions[name]}")

print("\nCIRCUIT DIAGRAMS")
for name, circuit in zip(names, circuits):
    print(f"\n{name}")
    print(circuit.draw(output="text"))

# 4. Execute each circuit for 100 shots and record counts.
counts_list = []
print("\nMEASUREMENT RESULTS (100 SHOTS EACH)")
for name, circuit in zip(names, circuits):
    job = simulator.run(circuit, shots=SHOTS)
    counts = job.result().get_counts()
    counts_list.append(counts)

    # Include zero-count outcomes explicitly for easier comparison.
    zeros = counts.get("0", 0)
    ones = counts.get("1", 0)

    print(f"\n{name}")
    print(f"Prediction: {predictions[name]}")
    print(f"Actual counts: {counts}")
    print(f"Outcome 0: {zeros}/100 ({zeros}%)")
    print(f"Outcome 1: {ones}/100 ({ones}%)")

# 5. Display a combined histogram.
plot_histogram(
    counts_list,
    legend=names,
    title="Quantum Circuit Measurement Results (100 Shots Each)"
)
plt.tight_layout()
plt.show()

# 6. Short comparison based on the actual counts.
print("\nCOMPARISON AND CONCLUSION")
for name, counts in zip(names, counts_list):
    zeros = counts.get("0", 0)
    ones = counts.get("1", 0)
    if name == "Circuit A (X)":
        print(f"{name}: predicted all 1; observed {ones} outcomes of 1 and {zeros} outcomes of 0.")
    elif name == "Circuit B (X, X)":
        print(f"{name}: predicted all 0; observed {zeros} outcomes of 0 and {ones} outcomes of 1.")
    else:
        print(f"{name}: predicted a roughly 50/50 split; observed {zeros} outcomes of 0 and {ones} outcomes of 1.")
        print("Circuit C is probabilistic, so the two counts need not be exactly equal.")

print("\nIdeal simulator expectation: Circuit A -> all 1; Circuit B -> all 0; Circuit C -> approximately half 0 and half 1.")
```
