 In the challenge 3, how a quantum circuit can produce correlated outcomes is
investigated. Now, design a circuit that switches between correlated and opposite measurement outcomes.
Your goal is to create two different two-qubit behaviors using only a small number of quantum gates.

```python
%pip install qiskit qiskit-aer matplotlib
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Simulator and number of shots
simulator = AerSimulator()
SHOTS = 100

# -----------------------------------
# PART A: CORRELATED OUTCOMES
# -----------------------------------
circuit_a = QuantumCircuit(2, 2)

circuit_a.h(0)          # Create superposition
circuit_a.cx(0, 1)      # Entangle the two qubits
circuit_a.measure([0, 1], [0, 1])

# -----------------------------------
# PART B: OPPOSITE OUTCOMES
# -----------------------------------
circuit_b = QuantumCircuit(2, 2)

circuit_b.h(0)          # Create superposition
circuit_b.cx(0, 1)      # Entangle the two qubits
circuit_b.x(1)          # Flip qubit 1
circuit_b.measure([0, 1], [0, 1])

# -----------------------------------
# PREDICTIONS BEFORE EXECUTION
# -----------------------------------
print("PREDICTIONS")
print("Part A: 00 and 11 should dominate.")
print("Part B: 01 and 10 should dominate.")
print("Each expected outcome should occur approximately")
print("50 times out of 100 shots.")

# Display circuits
print("\nPART A CIRCUIT")
print(circuit_a.draw(output="text"))

print("\nPART B CIRCUIT")
print(circuit_b.draw(output="text"))

# -----------------------------------
# RUN BOTH CIRCUITS
# -----------------------------------
result_a = simulator.run(
    circuit_a, shots=SHOTS
).result()

result_b = simulator.run(
    circuit_b, shots=SHOTS
).result()

counts_a = result_a.get_counts()
counts_b = result_b.get_counts()

# Print actual results
print("\nPART A: CORRELATED OUTCOMES")
print("Measurement counts:", counts_a)

print("\nPART B: OPPOSITE OUTCOMES")
print("Measurement counts:", counts_b)

# Show all possible outcomes, including zeros
print("\nDETAILED COMPARISON")
print("Outcome | Part A | Part B")
for outcome in ["00", "01", "10", "11"]:
    print(
        f"  {outcome}   |"
        f"   {counts_a.get(outcome, 0):3d}  |"
        f"   {counts_b.get(outcome, 0):3d}"
    )

# -----------------------------------
# PART C: HISTOGRAMS
# -----------------------------------
plot_histogram(
    [counts_a, counts_b],
    legend=["Part A: Correlated", "Part B: Opposite"],
    title="Comparison of Two-Qubit Outcomes (100 Shots Each)"
)
plt.show()

# Separate histograms
plot_histogram(counts_a, title="Part A: Correlated Outcomes")
plt.show()

plot_histogram(counts_b, title="Part B: Opposite Outcomes")
plt.show()

# -----------------------------------
# CONCLUSION
# -----------------------------------
print("\nCONCLUSION")
print("Part A: The qubits produce matching outcomes (00, 11).")
print("Part B: The qubits produce opposite outcomes (01, 10).")
print("The X gate flips qubit 1, changing the correlation.")
print("Both circuits use 2 qubits and 3 gates before measurement.")
```
